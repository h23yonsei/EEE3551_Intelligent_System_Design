```python
import cv2
import numpy as np
import serial


image = cv2.imread("assign2/lena.jpg",0)
img2 = cv2.imread("assign2/sample.jpg")
image = cv2.resize(image,(128,128))
img2 = cv2.resize(img2,(128,128))
cv2.imwrite('./assign2/original.bmp',image)

transmit_data = np.array(image)


print(transmit_data.size)
```

```python
test = serial.Serial("COM4",115200,timeout=1) # Baud Rate = 115,200 bps
test.write(transmit_data)
```

```python
import time

result=[]
start_time = time.time()

while(True):
    temp = int.from_bytes(test.read(),byteorder='big')
    if(temp == 0):
        print("hold")
        print(temp)
    else:
        result.append(temp)
        print(temp)
        break

print("*"*50)

for l in range(128 * 128):
# for l in range(8000):    
    result.append(int.from_bytes(test.read(),byteorder='big'))

    if l%1000 == 0:
        print(l)
    
```

```python
k=0
for i in range(128):
    for j in range(128):
        img2[i][j] = result[k]
        
        if (image[i][j] != result[k]):
            print(i, j, "to_FPGA :", image[i][j], "from_FPGA:" ,  result[k])
        k=k+1


cv2.imwrite('./assign2/result.bmp',img2)

print(result)
```
